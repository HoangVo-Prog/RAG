```python
%cd /kaggle/working/RAG/aio-rag
import subprocess
import time
import requests
from pyngrok import ngrok

def wait_for_ui_ready(port=8501, timeout=600):
    url = f"http://localhost:{port}"
    print(f"⏳ Waiting for Streamlit UI at {url}")
    start_time = time.time()
    while time.time() - start_time < timeout:
        try:
            r = requests.get(url)
            if "streamlit" in r.text.lower() or "streamlitApp" in r.text:
                print("✅ Streamlit is ready!")
                return True
        except:
            pass
        time.sleep(5)
    raise TimeoutError("❌ Streamlit not ready in time.")

# 🔥 Step 1: Start Streamlit in background (cross-platform-safe)
print("🚀 Launching Streamlit...")
streamlit_proc = subprocess.Popen(
    ["streamlit", "run", "main.py", "--server.port", "8501"],
    stdout=subprocess.PIPE,
    stderr=subprocess.PIPE
)

# 🔄 Step 2: Wait for Streamlit UI to be ready
wait_for_ui_ready()

# 🌍 Step 3: Connect ngrok
public_url = ngrok.connect("http://localhost:8501")
print("🌐 Ngrok Public URL:", public_url)
```
